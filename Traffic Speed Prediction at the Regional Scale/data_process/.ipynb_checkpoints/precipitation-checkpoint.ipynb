```python
import pandas as pd
import numpy as np
import os
from glob import glob
from pathlib import Path

# ---------- locate project root ----------
def find_project_root():
    p = Path.cwd().resolve()
    for parent in [p] + list(p.parents):
        if (parent / "Main data").exists():
            return parent
    raise RuntimeError("Project root (CODE UPLOAD) not found.")

PROJECT_ROOT = find_project_root()

# ---------- INPUT: precipitation CSV folder (relative) ----------
input_dir = (
    PROJECT_ROOT
    / "Main data"
)

# ---------- OUTPUT: prediction_data/precipitation (relative) ----------
output_dir = PROJECT_ROOT / "Main data" / "prediction_data" / "precipitation"
output_dir.mkdir(parents=True, exist_ok=True)

# ---------- CSV files ----------
csv_files = [
    input_dir / "徐家汇.csv",
    input_dir / "浦东.csv"
]

# 若后续代码需要字符串路径，可显式转换
csv_files = [str(p) for p in csv_files]

csv_files
```

```python
# 处理每个文件
for i, file_path in enumerate(csv_files, start=1):
    try:
        # if i == 2:
        #     break
        # 读取数据
        df = pd.read_csv(file_path)

        # 检查是否包含目标列
        if "降水量（mm）" not in df.columns:
            print(f"⚠️ 跳过文件，未找到 '降水量（mm）' 列：{file_path}")
            continue

        # 保留原始列，不复制
        repeated_df = df[["降水量（mm）"]].copy()

        # 保存为 station{i}.csv
        output_path = os.path.join(output_dir, f"station{i}.csv")
        repeated_df.to_csv(output_path, index=False, header=False, encoding='utf-8-sig')
        print(f"✅ 已保存: {output_path}")

    except Exception as e:
        print(f"❌ 处理失败 {file_path}: {e}")
```
