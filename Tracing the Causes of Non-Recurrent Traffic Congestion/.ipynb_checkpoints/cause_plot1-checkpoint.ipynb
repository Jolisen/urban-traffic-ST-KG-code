```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# =========================
# ✅ 期刊字体统一设置（新增）
# =========================
mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "DejaVu Sans"],
    "font.size": 40,          # 基础字号（与你原来 xtick/label 一致）
    "axes.titlesize": 35,
    "legend.fontsize": 24,
    "legend.title_fontsize": 30,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

# =========================
# 原始代码（未改动逻辑）
# =========================

# 转为 DataFrame
df = pd.DataFrame(data_list)

# 拥堵等级数值映射
level_map = {
    'smooth': 2,
    'light_congestion': 3,
    'congestion': 4,
    'severe_congestion': 5
}

# 所有唯一时间段（按升序）
batches = sorted(df['batch'].unique())
n_batches = len(batches)
batch_to_angle = {batch: i * 2 * np.pi / n_batches for i, batch in enumerate(batches)}

# 数据透视表
pivot = df.pivot_table(
    index='honeycomb_id',
    columns='batch',
    values='grid_congestion_level',
    aggfunc='first'
)
pivot = pivot.reindex(columns=batches).fillna('nan')
pivot = pivot.applymap(lambda x: level_map.get(x, 1))

# 👇 排除某个 ID
pivot = pivot[pivot.index != 92698]

# 所有 honeycomb_id
honeycomb_ids = pivot.index.tolist()

# 如果存在 93551，就把它排到倒数第二位
if 93551 in honeycomb_ids:
    honeycomb_ids.remove(93551)
    honeycomb_ids.insert(-1, 93551)  # 倒数第二个位置

# 分组绘图：每3个为一组
group_size = 3
num_groups = (len(honeycomb_ids) + group_size - 1) // group_size

for group_idx in range(num_groups):
    group_ids = honeycomb_ids[group_idx * group_size: (group_idx + 1) * group_size]
    colors = plt.cm.viridis(np.linspace(0, 0.6, len(group_ids)))

    fig, ax = plt.subplots(
        figsize=(14, 14),
        subplot_kw=dict(polar=True)
    )

    for i, idx in enumerate(group_ids):
        row = pivot.loc[idx]
        values = row.values.tolist()
        angles = list(batch_to_angle.values())

        # 闭合图形
        values += [values[0]]
        angles += [angles[0]]

        ax.plot(angles, values, label=f'ID {idx}', color=colors[i])
        ax.fill(angles, values, alpha=0.1, color=colors[i])

    # 坐标样式
    ax.set_theta_zero_location("N")
    ax.set_theta_direction(-1)
    ax.set_xticks(list(batch_to_angle.values()))
    ax.set_xticklabels(batches, rotation=90, fontsize=24)
    ax.tick_params(axis='x', pad=10)

    # 半径标签
    r_levels = [1, 2, 3, 4, 5]
    r_labels = [
        'No Data',
        'Smooth',
        'Light Congestion',
        'Congestion',
        'Severe Congestion'
    ]
    ax.set_yticks(r_levels)
    ax.set_yticklabels([])

    for r, label in zip(r_levels, r_labels):
        ax.text(
            112.5,
            r,
            label,
            ha='center',
            va='center',
            fontsize=24,
            color='black',
            rotation=0
        )

    ax.legend(
        loc='upper right',
        bbox_to_anchor=(1.25, 1.1),
        fontsize=24,
        title='Grid ID',
        title_fontsize=30
    )

    plt.tight_layout()

    # =========================
    # ✅ 输出分辨率改为 1200 dpi
    # =========================
    figures_dir = PROJECT_ROOT / "figures" / "grid_radar"
    figures_dir.mkdir(parents=True, exist_ok=True)
    
    out_path = figures_dir / f"grid_radar_group_{group_idx + 1}.tif"
    plt.savefig(out_path, dpi=1200, bbox_inches="tight")
    print(f"✅ Saved: {out_path}")

    plt.show()
```

```python
if honeycomb_ids_ls is not None:
    honeycomb_ids_str = ', '.join(f"{honeycomb_id}" for honeycomb_id in honeycomb_ids_ls)
    query = f"""
            MATCH (h:honeycomb)
            WHERE h.name IN [{honeycomb_ids_str}]
            RETURN h.name AS honeycomb_id, h.geometry AS honeycomb_geometry
            """
    honeycomb_ids_result = g.run(query).data()
    honeycomb_ids_resultdf = pd.DataFrame(honeycomb_ids_result)
    honeycomb_ids_resultdf['honeycomb_geometry'] = honeycomb_ids_resultdf['honeycomb_geometry'].apply(wkt.loads)
    honeycomb_ids_result_gdf = gpd.GeoDataFrame(honeycomb_ids_resultdf, geometry='honeycomb_geometry', crs=roadcrs)
if road_ids_ls is not None:
    road_ids_str = ', '.join(f"{road_id}" for road_id in road_ids_ls)
    query = f"""
            MATCH (r:road)
            WHERE r.name IN [{road_ids_str}]
            RETURN r.name AS road_id, r.roadname AS road_name, r.geometry AS road_geometry
            """
    road_ids_result = g.run(query).data()
    road_ids_resultdf = pd.DataFrame(road_ids_result)
    road_ids_resultdf['road_geometry'] = road_ids_resultdf['road_geometry'].apply(wkt.loads)
    road_ids_result_gdf = gpd.GeoDataFrame(road_ids_resultdf, geometry='road_geometry', crs=roadcrs)
    
    
if poi_ids_ls is not None:
    poi_ids_str = ', '.join(f"{poi_id}" for poi_id in poi_ids_ls)
    query = f"""
            MATCH (p:POI)
            WHERE p.name IN [{poi_ids_str}]
            RETURN p.name AS poi_id, p.poi_name AS poi_name, p.geometry AS poi_geometry
            """
    poi_ids_result = g.run(query).data()
    poi_ids_resultdf = pd.DataFrame(poi_ids_result)
    poi_ids_resultdf['poi_geometry'] = poi_ids_resultdf['poi_geometry'].apply(wkt.loads)
    poi_ids_result_gdf = gpd.GeoDataFrame(poi_ids_resultdf, geometry='poi_geometry', crs=roadcrs)

print("honeycomb_ids_result_gdf\n", honeycomb_ids_result_gdf)
print("road_ids_result_gdf\n", road_ids_result_gdf)
print("poi_ids_result_gdf\n", poi_ids_result_gdf)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# =========================
# ✅ 同款字体与字号设置（与前一段完全一致）
# =========================
mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "DejaVu Sans"],
    "font.size": 40,          # 基础字号
    "axes.titlesize": 35,
    "legend.fontsize": 24,
    "legend.title_fontsize": 30,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

# =========================
# 原始逻辑（不改）
# =========================

# 拥堵等级映射
level_map = {
    'smooth': 2,
    'light_congestion': 3,
    'congestion': 4,
    'severe_congestion': 5
}

# 准备数据
filtered_expanded_road = expanded_road[
    expanded_road['road_id'].isin(roads_in_grid['road_id'])
].copy()

filtered_expanded_road['level'] = (
    filtered_expanded_road['road_congestion_level']
    .map(level_map)
    .fillna(1)
)

# 获取唯一时间（batch）和 road_id
batches = sorted(filtered_expanded_road['batch'].unique())
road_ids = sorted(filtered_expanded_road['road_id'].unique())

print(f"road_ids:\n{road_ids}")
print(f"n_roads:\n{len(road_ids)}")

# 将 road_id 均分为 3 组
groups = np.array_split(road_ids, 3)

# 角度映射
angles = np.linspace(0, 2 * np.pi, len(batches), endpoint=False).tolist()
angle_map = dict(zip(batches, angles))

# =========================
# 绘制雷达图
# =========================
for i, group in enumerate(groups, start=1):

    # ✅ figsize 改为 (14, 14)
    fig, ax = plt.subplots(
        figsize=(14, 14),
        subplot_kw=dict(polar=True)
    )

    colors = plt.cm.viridis(np.linspace(0, 0.6, len(group)))

    for j, road_id in enumerate(group):
        sub_df = filtered_expanded_road[
            filtered_expanded_road['road_id'] == road_id
        ]

        level_series = pd.Series(index=batches, data=1)
        level_series[sub_df['batch']] = sub_df['level'].values

        values = level_series.tolist()
        values += [values[0]]

        angle_list = [angle_map[b] for b in batches]
        angle_list += [angle_list[0]]

        ax.plot(
            angle_list,
            values,
            label=f'ID {road_id}',
            linewidth=1.5,
            color=colors[j]
        )
        ax.fill(angle_list, values, alpha=0.1, color=colors[j])

    # 雷达图坐标设置
    ax.set_theta_zero_location("N")
    ax.set_theta_direction(-1)

    ax.set_xticks(angles)
    ax.set_xticklabels(batches, fontsize=24, rotation=90)
    ax.tick_params(axis='x', pad=10)

    # y 轴环标签
    r_levels = [1, 2, 3, 4, 5]
    r_labels = [
        'No Data',
        'Smooth',
        'Light Congestion',
        'Congestion',
        'Severe Congestion'
    ]
    ax.set_yticks(r_levels)
    ax.set_yticklabels([])

    for r, label in zip(r_levels, r_labels):
        ax.text(
            112.5,
            r,
            label,
            ha='center',
            va='center',
            fontsize=24,
            color='black'
        )

    # =========================
    # ✅ legend：与你指定的完全一致
    # =========================
    ax.legend(
        loc='upper right',
        bbox_to_anchor=(1.25, 1.1),
        fontsize=24,
        title='Road ID',
        title_fontsize=30
    )

    plt.tight_layout()

    # =========================
    # ✅ 1200 dpi 输出
    # =========================
    figures_dir = PROJECT_ROOT / "figures" / "road_radar"
    figures_dir.mkdir(parents=True, exist_ok=True)
    
    out_path = figures_dir / f"road_radar_group_{i}.tif"
    plt.savefig(out_path, dpi=1200, bbox_inches="tight")

    plt.show()
```
